2. Exécute les cellules dans l'ordre (▶).
3. Récupère le ZIP à la fin, puis ouvre-le dans MusiMix avec **➕ ZIP**.

Compte ~2-4 min par morceau sur GPU.

```
#@title 2. Envoie ton fichier audio (et, si tu l'as, la grille / les paroles en .txt)
from google.colab import files
up = files.upload()
names = list(up)
audio = [n for n in names if n.lower().endswith(('.mp3','.wav','.flac','.m4a','.ogg'))][0]
texte = next((n for n in names if n.lower().endswith(('.txt','.cho'))), None)
print('audio :', audio, '| texte :', texte)
```

```
#@title 3. Réglages puis lancement
langue = 'pt'  #@param ['fr','pt','en','es','it','de']
capo = 0  #@param {type:'integer'}
titre = ''  #@param {type:'string'}
artiste = ''  #@param {type:'string'}
pistes = 'htdemucs_6s'  #@param ['htdemucs_6s','htdemucs','two']
texte_est = 'grille (accords + paroles)'  #@param ['grille (accords + paroles)','paroles seules']
whisper = 'large-v3'  #@param ['medium','large-v3']

args = [audio, '--lang', langue, '--model', pistes, '--whisper-model', whisper, '--out', 'packs']
if capo: args += ['--capo', str(capo)]
if titre: args += ['--title', titre]
if artiste: args += ['--artist', artiste]
if texte: args += ['--chords' if texte_est.startswith('grille') else '--lyrics', texte]
import shlex
!python musimix_prep.py {' '.join(shlex.quote(a) for a in args)}
```

```
#@title 4. Télécharger le pack
import glob
for z in glob.glob('packs/*.zip'):
    files.download(z)
```
